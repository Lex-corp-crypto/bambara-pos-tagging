# ENV et OUTIS

```python
import re
import torch
import torch.nn as nn

# Exemple initial de données
training_data = [
    ("Amadou étudie le bambara".split(), ["NOM", "VERBE", "DET", "NOM"]),
    ("le modèle comprend les phrases".split(), ["DET", "NOM", "VERBE", "DET", "NOM"])
]
```

<a id="vocab-mapping"></a>

## 2. Création des Dictionnaires (Vocabulaire et Tags)

```python
# Dictionnaire des mots avec gestion de <PAD> et <UNK>
word_to_ix = {"<PAD>": 0, "<UNK>": 1}
for sentence, tags in training_data:
    for word in sentence:
        if word not in word_to_ix:
            word_to_ix[word] = len(word_to_ix)

# Dictionnaire des étiquettes (Tags)
tag_to_ix = {"<PAD>": 0, "NOM": 1, "VERBE": 2, "DET": 3}

print("Dictionnaire des mots (word_to_ix) :")
print(word_to_ix)
print("\nDictionnaire des étiquettes (tag_to_ix) :")
print(tag_to_ix)
```

# <a id="prepare-sequence"></a>
## 3. Conversion de Séquences de Mots en Tenseurs PyTorch

```python
def prepare_sequence(seq, to_ix):
    """
    Convertit une liste de mots/tags en un tenseur PyTorch d'indices.
    Utilise le token <UNK> pour les mots inconnus au vocabulaire.
    """
    idxs = []
    for w in seq:
        if w in to_ix:
            idxs.append(to_ix[w])
        else:
            idxs.append(to_ix["<UNK>"])
    return torch.tensor(idxs, dtype=torch.long)

# Test de la fonction
exemple_phrase = "Amadou comprend le bambara".split()
tensor_phrase = prepare_sequence(exemple_phrase, word_to_ix)

print("Phrase d'origine :", exemple_phrase)
print("Tenseur PyTorch généré :", tensor_phrase)
```

# <a id="tokenization"></a>
## 4. Comparaison et Adaptation de la Tokenisation (Français vs Bambara)

```python
def bamanankan_tokenizer(text):
    """
    Tokeniseur adapté pour le Bambara (Bamanankan).
    Gère les caractères spécifiques (ɛ, ɔ, ɲ, ŋ) et isole la ponctuation.
    """
    text = text.strip()
    pattern = r"[a-zA-ZÀ-ÿɛƐɔƆɲƝŋŊ]+|[.,!\?\(\):;]"
    return re.findall(pattern, text)

# Test du tokeniseur Bambara
phrase_bambara = "Amadou bɛ kalan kɛ, a bɛ bamanankan fɔ ɲuman !"
tokens = bamanankan_tokenizer(phrase_bambara)
print("Tokens Bambara générés :", tokens)
```

# 5. Prétraitement — Tokenisation du Bambara

```python
import re

def bamanankan_tokenizer(text):
    """
    Tokenise une phrase en bambara.
    Capture les mots (lettres latines + accents français + caractères bambara)
    et isole la ponctuation comme token séparé.
    """
    text = text.strip()
    pattern = r"[a-zA-ZÀ-ÿɛƐɔƆɲƝŋŊ]+|[.,!?'\";:()\[\]]"
    return re.findall(pattern, text)

# Test rapide
phrase = "Amadou bɛ kalan kɛ, a bɛ bamanankan fɔ ɲuman !"
print(bamanankan_tokenizer(phrase))
```
